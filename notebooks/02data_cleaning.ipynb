###Selecting only necessary columns

```python
cols=['refYear', 'partnerDesc', 'cmdCode', 'cifvalue']
df_clean=df[cols].copy()
df_clean.head()
```

```python
df_clean.columns= ['Year', 'Partner', 'Code', 'USD_Value_bn']
df_clean= df_clean.replace('Rep. of Korea', 'South Korea')
df_clean.head()
```

```python
df_clean['USD_Value_bn']= (df_clean['USD_Value_bn']/1e9)
df_clean.head()
```

```python
total=df_clean.groupby(['Year','Partner'])['USD_Value_bn'].sum().reset_index()
```

###Creating pivot tables by country

```python
pivot=total.pivot(index='Year', columns='Partner', values='USD_Value_bn').fillna(0)
pivot['total']= pivot.sum(axis=1)
pivot['S.Korea_YoY_change_%']= pivot['South Korea'].pct_change().mul(100).round(2)
pivot['China_YoY_change_%']= pivot['China'].pct_change().mul(100).round(2)
pivot['Singapore_YoY_change_%']= pivot['Singapore'].pct_change().mul(100).round(2)
pivot['Vietnam_YoY_change_%']= pivot['Viet Nam'].pct_change().mul(100).round(2)
pivot['USA_YoY_change_%']= pivot['USA'].pct_change().mul(100).round(2)
```

###Exporting the data

```python
pivot.to_excel('TradeDataWhole_cleaned.xlsx')
```
